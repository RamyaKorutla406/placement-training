```python
l=[1,2,3,4,5,7]
flag=0
for i in range(len(l)-1):
  if l[i]>=l[i+1]:
    flag=1
    break
if flag==1:
  print("Nope")
else:
  print("Yes")
```

```python
l=[4,3,4,5,1,3,2,1,5,2,3,5]
k=4
sum=0
for i in range(k):
  sum+=l[i]
m=sum
for i in range(len(l)-k):
  sum=sum+l[i+k]-l[i]
  m=max(m,sum)
print(m)
```

```python
a=[2,5,1,7,10]
k=14
l,r,sum,m=0,0,0,0
while r<len(a):
    sum+=a[r]
    while sum>=k:
      sum-=a[l]
      l+=1
    if sum<k:
      m=max(m,(r-l)+l)
    r+=1
print(m)
```

```python
s="ababab"
m = 0
for i in range(len(s)):
    l, r = i, i
    while l >= 0 and r < len(s) and s[l] == s[r]:
      m = max(m, r - l + 1)
      l -= 1
      r += 1
for i in range(len(s)-1):
    l, r = i, i + 1
    while l >= 0 and r < len(s) and s[l] == s[r]:
      m = max(m, r - l + 1)
      l -= 1
      r += 1
print(m)
```

```python
s = "ababab"
count = 0
for i in range(len(s)):
    l = i
    r = i
    while l >= 0 and r < len(s) and s[l] == s[r]:
        count += 1
        l -= 1
        r += 1
    l = i
    r = i + 1
    while l >= 0 and r < len(s) and s[l] == s[r]:
        count += 1
        l -= 1
        r += 1

print(count)
```

```python
arr=[900,945,955,1100,1500,1800]
dep=[920,1200,1130,1150,1900,2000]
arr.sort()
dep.sort()
i,j,doct,m=1,0,1,0
while i<len(arr) and j<len(dep):
  if arr[i]<=dep[j]:
    doct+=1
    i+=1
  else:
    doct-=1
    j+=1
  m=max(m,doct)
print(m)
```
