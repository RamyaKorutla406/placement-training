```python
l1=[1,4,5,7,9,10]
l2=[2,3,6,8]
res=[]
i=0
j=0
while i<len(l1) and j<len(l2):
  if l1[i]<l2[j]:
    res.append(l1[i])
    i+=1
  else:
    res.append(l2[j])
    j+=1
res.extend(l1[i:])
res.extend(l2[j:])
print(res)
```

```python
def div(arr):
  if len(arr)<=1:
     return arr
  mid=len(arr)//2
  left=div(arr[:mid])
  right=div(arr[mid:])
  return merge(left,right)
def merge(left,right):
   res=[]
   i=0
   j=0
   while(i<len(left) and j<len(right)):
     if left[i]<right[j]:
       res.append(left[i])
       i+=1
     else:
       res.append(right[j])
       j+=1
   res.extend(left[i:])
   res.extend(right[j:])
   print(res)
```
