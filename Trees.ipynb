```python
def count(arr):
  result = []
  for i in range(len(arr)):
    count = 1
    for j in range(i):
      if arr[j] <= arr[i]:
        count += 1
    result.append(count)
  return result
arr= [100, 50, 60, 70, 60, 75, 80, 85]
counts = count(arr)
print(counts)
```

```python
class Node:
  def __init__(self,key):
    self.l=None
    self.r=None
    self.value=key

def insert(root,key):
  if root is None:
    return Node(key)
  else:
    if key<root.value:
      root.l=insert(root.l,key)
    else:
      root.r=insert(root.r,key)
    return root

def inorder(root):
  if root:
    inorder(root.l)
    print(root.value)
    inorder(root.r)
r=Node(7)
r=insert(r,5)
r=insert(r,3)
r=insert(r,6)
r=insert(r,2)
r=insert(r,4)
r=insert(r,1)

inorder(r)
print(r)
```

```python
def postOrder(root):
    if root:
        postOrder(root.left)
        postOrder(root.right)
        print(root.info, end=" ")
```

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None
  def top_view(root):
    q = []
    d = dict()
    q.append((root, 0))
    while q:
      node, val = q.pop(0)
      if val not in d:
        d[val] = node.data
      if node.left:
        q.append((node.left, val - 1))
      if node.right:
        q.append((node.right, val + 1))
    for i in sorted(d):
      print(d[i], end=' ')
root = Node(5)
root.left = Node(2)
root.right = Node(7)
root.left.left = Node(1)
root.left.right = Node(3)
root.right.left = Node(6)
root.right.right = Node(8)
Node.top_view(root)
```

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None

  def right_view(root):
    if root is None:
      return
    q = []
    d = dict()
    q.append((root, 0))
    while q:
      node, level = q.pop(0)
      if level not in d:
        d[level] = node.data
      if node.right:
        q.append((node.right, level + 1))
      if node.left:
        q.append((node.left, level + 1))
    for i in sorted(d):
      print(d[i], end=' ')

root = Node(5)
root.left = Node(2)
root.right = Node(7)
root.left.left = Node(1)
root.left.right = Node(3)
root.right.left = Node(6)
root.right.right = Node(8)

Node.right_view(root)
```
