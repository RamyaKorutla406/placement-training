```python
def fib(n):
  if(n==0 or n==1):
    return n
  return fib(n-1)+fib(n-2)
n=int(input())
print(fib(n))
```

```python
n=6
memo=[-1]*(n+1)
def fib(n):
  if n<=1:
    return n
  if memo[n]!=-1:
    return memo[n]
  memo[n]=fib(n-1)+fib(n-2)
  return memo[n]
print(fib(n))
```

```python
#tabulation
n=6
dp=[-1]*(n+1)
dp[1]=1
dp[0]=0
for i in range(2,n+1):
    dp[i]=dp[i-1]+dp[i-2]
print(dp[n])
```

```python
rock=[30,10,60,10,60,50]
def frog(ind):
  if ind==0:
    return 0
  j1=frog(ind-1)+abs(rock[ind]-rock[ind-1])
  if ind>1:
    j2=frog(ind-2)+abs(rock[ind]-rock[ind-2])
    return min(j1,j2)
  else:
    return j1
print(frog(len(rock)-1))
```

```python
#frog tabulation
dp=[-1]*len(rock)
def frog(ind):
  if ind==0:
    return 0
  if dp[ind]!=-1:
    return dp[ind]
  j1=frog(ind-1)+abs(rock[ind]-rock[ind-1])
  if ind>1:
    j2=frog(ind-2)+abs(rock[ind]-rock[ind-2])
    dp[ind]=min(j1,j2)
    return dp[ind]
  else:
    return j1
print(frog(len(rock)-1))
```

```python
rock=[30,10,60,10,60,50]
n=6
dp=[0]*n
for i in range(1,n):
  j1=dp[i-1]+abs(rock[i]-rock[i-1])
  j2=float('inf')
  if i>1:
    j2=dp[i-2]+abs(rock[i]-rock[i-2])
    dp[i]=min(j1,j2)
print(dp[n-1])
```

```python
arr=[1,2,5,10]
target=12
n=len(arr)
dp=[[False]*(target+1) for _ in range(n+1)]
for i in range(n+1):
  dp[i][0]=True
for i in range(1,n+1):
  for j in range(1,target+1):
    if arr[i-1]>j:
      dp[i][j]=dp[i-1][j]
    else:
      dp[i][j]=dp[i-1][j] or dp[i-1][j-arr[i-1]]
if dp[n][target]:
  print("Yes. can make an amount")
else:
  print("No. can't make an amount")
```
